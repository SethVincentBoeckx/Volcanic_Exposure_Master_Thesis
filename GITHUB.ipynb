# Countries

## POINT

```python
PROJECT_ID = "your-gee-project-id"

VOLCANO_CSV_PATH = "data/volcanoes_africa.csv"
OUTPUT_FILE = "output/population.xlsx"
CHECKPOINT_CSV = "output/population_checkpoint.csv"

ASSET_STD = f"projects/{PROJECT_ID}/assets/GADM_410"
COUNTRY_COLUMN = "COUNTRY"
GHS_BASE_ASSET_ID = f"projects/{PROJECT_ID}/assets/WGS_"
BAND = "b1"
BUFFER_DISTANCES = [10000, 30000, 100000]
OVERSAMPLE = 10

ALL_YEARS = list(range(1975, 2031, 5)) + list(range(2040, 2101, 10))

COUNTRY_LIST = sorted([
    "Cabo Verde", "Cameroon", "Chad", "Comoros", "Democratic Republic of the Congo",
    "Djibouti", "Equatorial Guinea", "Eritrea", "Ethiopia", "Mayotte",
    "Kenya", "Malawi", "Niger", "Nigeria", "Rwanda", "South Sudan", "Sudan",
    "Uganda", "Tanzania", "Zambia",
])


def load_volcanoes_geometry(csv_path):
    with open(csv_path, "r", encoding="utf-8-sig") as f:
        first_line = f.readline()
    delimiter = ";" if first_line.count(";") > first_line.count(",") else ","
    df = pd.read_csv(csv_path, delimiter=delimiter, encoding="utf-8-sig")
    df.columns = df.columns.str.strip()
    df = df.dropna(subset=["Latitude", "Longitude"])
    features = [
        ee.Feature(ee.Geometry.Point([
            float(str(row["Longitude"]).replace(",", ".")),
            float(str(row["Latitude"]).replace(",", "."))
        ]))
        for _, row in df.iterrows()
    ]
    return ee.FeatureCollection(features)


def make_fractional_mask(geometry, projection, scale):
    return (
        ee.Image.constant(1)
        .clip(geometry)
        .unmask(0)
        .reproject(crs=projection, scale=scale / OVERSAMPLE)
        .reduceResolution(reducer=ee.Reducer.mean(), maxPixels=OVERSAMPLE * OVERSAMPLE * 4)
        .reproject(crs=projection, scale=scale)
    )


def build_excel(checkpoint_path, output_path):
    df_raw = pd.read_csv(checkpoint_path)
    if df_raw.empty:
        return

    final_list = []
    for country in df_raw["Country"].unique():
        for r_km in [d // 1000 for d in BUFFER_DISTANCES]:
            subset = df_raw[
                (df_raw["Country"] == country) & (df_raw["Radius"] == r_km)
            ].copy().sort_values("Year").reset_index(drop=True)
            if subset.empty:
                continue

            subset["PopAn"] = subset["PopA"] / subset["PopT"].replace(0, np.nan)
            final_list.append(subset)

    df_final = pd.concat(final_list, ignore_index=True)

    with pd.ExcelWriter(output_path, engine="xlsxwriter") as writer:
        for sheet_name, col in [
            ("PopA_Absolute", "PopA"),
            ("PopT_Absolute", "PopT"),
            ("PopAn_Share", "PopAn"),
        ]:
            pivot = df_final.pivot_table(
                index=["Country", "Radius"], columns="Year", values=col, aggfunc="mean"
            ).reset_index()
            pivot.columns.name = None
            pivot.to_excel(writer, sheet_name=sheet_name, index=False)

        for sheet in writer.sheets.values():
            sheet.set_column("A:A", 35)
            sheet.set_column("B:B", 10)
            sheet.set_column("C:Z", 12)


def load_checkpoint_years():
    completed_years = set()
    if os.path.exists(CHECKPOINT_CSV):
        df_ck = pd.read_csv(CHECKPOINT_CSV)
        year_counts = df_ck.groupby("Year")["Country"].nunique()
        completed_years = set(year_counts[year_counts >= len(COUNTRY_LIST)].index.tolist())
    return completed_years


def run_analysis():
    ee.Initialize(project=PROJECT_ID)

    fc_countries = ee.FeatureCollection(ASSET_STD).filter(
        ee.Filter.inList(COUNTRY_COLUMN, COUNTRY_LIST)
    )
    volcano_fc = load_volcanoes_geometry(VOLCANO_CSV_PATH)

    ref_img = ee.Image(f"{GHS_BASE_ASSET_ID}1975").select(BAND)
    native_proj = ref_img.projection()
    native_scale = native_proj.nominalScale().getInfo()

    buffer_fractions = {}
    for radius in BUFFER_DISTANCES:
        buffer_geom = volcano_fc.map(lambda f: f.buffer(radius)).union().geometry()
        buffer_fractions[radius] = make_fractional_mask(buffer_geom, native_proj, native_scale)

    completed_years = load_checkpoint_years()

    for year in ALL_YEARS:
        if year in completed_years:
            continue

        try:
            raw_img = ee.Image(f"{GHS_BASE_ASSET_ID}{year}").select(BAND)
            img = raw_img.toDouble().where(raw_img.lt(0), 0).unmask(0)

            combined = img.rename("PopT")
            for radius in BUFFER_DISTANCES:
                r_km = radius // 1000
                combined = combined.addBands(
                    img.multiply(buffer_fractions[radius]).rename(f"PopA_{r_km}")
                )

            result_fc = combined.reduceRegions(
                collection=fc_countries,
                reducer=ee.Reducer.sum(),
                scale=native_scale,
                tileScale=16,
            )
            features = result_fc.getInfo()["features"]

            year_records = []
            for feat in features:
                props = feat["properties"]
                country_name = props.get(COUNTRY_COLUMN)
                if country_name not in COUNTRY_LIST:
                    continue
                pop_t = max(0.0, float(props.get("PopT") or 0.0))
                for radius in BUFFER_DISTANCES:
                    r_km = radius // 1000
                    pop_a = max(0.0, float(props.get(f"PopA_{r_km}") or 0.0))
                    year_records.append({
                        "Country": country_name,
                        "Year": year,
                        "Radius": r_km,
                        "PopA": pop_a,
                        "PopT": pop_t,
                    })

            df_new = pd.DataFrame(year_records)
            file_exists = os.path.exists(CHECKPOINT_CSV)
            df_new.to_csv(CHECKPOINT_CSV, mode="a", header=not file_exists, index=False)

        except Exception as e:
            print(f"Failed year {year}: {e}")
            time.sleep(3)

    if os.path.exists(CHECKPOINT_CSV):
        build_excel(CHECKPOINT_CSV, OUTPUT_FILE)


if __name__ == "__main__":
    run_analysis()
```

## Polygon

```python
PROJECT_ID = "your-gee-project-id"
VOLCANO_ASSET_PATH = f"projects/{PROJECT_ID}/assets/volcano_polygons"
ASSET_STD = f"projects/{PROJECT_ID}/assets/GADM_410"
COUNTRY_COLUMN = "COUNTRY"

OUTPUT_FILE = "output/population_polygons.xlsx"
CHECKPOINT_CSV = "output/population_polygons_checkpoint.csv"

GHS_BASE_ASSET_ID = f"projects/{PROJECT_ID}/assets/WGS_"
BAND = "b1"
BUFFER_DISTANCES = [10000, 30000, 100000]
ALL_YEARS = list(range(1975, 2031, 5)) + list(range(2040, 2101, 10))

COUNTRY_LIST = sorted([
    "Cabo Verde", "Cameroon", "Chad", "Comoros", "Democratic Republic of the Congo",
    "Djibouti", "Equatorial Guinea", "Eritrea", "Ethiopia", "Mayotte",
    "Kenya", "Malawi", "Niger", "Nigeria", "Rwanda", "South Sudan", "Somalia", "Sudan",
    "Uganda", "Tanzania", "Zambia",
])


def load_volcanoes_geometry(asset_path):
    volcano_fc = ee.FeatureCollection(asset_path)
    asset_features = volcano_fc.getInfo()["features"]
    hulls = [
        ee.Feature(feat).geometry().convexHull(maxError=100)
        for feat in asset_features
    ]
    return ee.FeatureCollection([ee.Feature(h) for h in hulls])


def build_excel_from_csv():
    if not os.path.exists(CHECKPOINT_CSV):
        return
    df_raw = pd.read_csv(CHECKPOINT_CSV)
    if df_raw.empty:
        return

    final_list = []
    for country in df_raw["Country"].unique():
        for r_km in [d // 1000 for d in BUFFER_DISTANCES]:
            subset = df_raw[
                (df_raw["Country"] == country) & (df_raw["Radius"] == r_km)
            ].copy().sort_values("Year").reset_index(drop=True)
            if subset.empty:
                continue
            subset["PopAn"] = subset["PopA"] / subset["PopT"].replace(0, np.nan)
            final_list.append(subset)

    df_final = pd.concat(final_list, ignore_index=True)

    def make_pivot(df, r_km, value_col):
        subset = df[df["Radius"] == r_km]
        pivot = subset.pivot_table(
            index="Country", columns="Year", values=value_col, aggfunc="mean"
        ).reset_index()
        pivot.columns.name = None
        return pivot

    with pd.ExcelWriter(OUTPUT_FILE, engine="xlsxwriter") as writer:
        for r_km in [d // 1000 for d in BUFFER_DISTANCES]:
            for sheet_suffix, col in [
                ("PopA", "PopA"),
                ("PopAn", "PopAn"),
            ]:
                pivot = make_pivot(df_final, r_km, col)
                pivot.to_excel(writer, sheet_name=f"{r_km}km_{sheet_suffix}", index=False)

        pivot = df_final.pivot_table(
            index=["Country", "Radius"], columns="Year", values="PopT", aggfunc="mean"
        ).reset_index()
        pivot.columns.name = None
        pivot.to_excel(writer, sheet_name="PopT_Absolute", index=False)

        for sheet in writer.sheets.values():
            sheet.set_column("A:A", 35)
            sheet.set_column("B:Z", 12)


def load_checkpoint_years():
    completed_years = set()
    if os.path.exists(CHECKPOINT_CSV):
        df_existing = pd.read_csv(CHECKPOINT_CSV)
        if not df_existing.empty:
            year_counts = df_existing.groupby("Year")["Country"].nunique()
            completed_years = set(year_counts[year_counts >= len(COUNTRY_LIST)].index.tolist())
    return completed_years


def run_analysis():
    ee.Initialize(project=PROJECT_ID)

    fc_countries_raw = ee.FeatureCollection(ASSET_STD).filter(
        ee.Filter.inList(COUNTRY_COLUMN, COUNTRY_LIST)
    )

    country_index = {name: i for i, name in enumerate(COUNTRY_LIST)}

    def add_index(feat):
        name = feat.get(COUNTRY_COLUMN)
        idx = ee.Dictionary(country_index).getNumber(name)
        return feat.set("country_idx", idx)

    fc_countries = fc_countries_raw.map(add_index)

    volcano_fc = load_volcanoes_geometry(VOLCANO_ASSET_PATH)

    buffer_masks = {}
    for radius in BUFFER_DISTANCES:
        buffered = volcano_fc.map(lambda f: f.buffer(radius))
        buffer_masks[radius] = ee.Image(0).paint(buffered, 1).gt(0)

    completed_years = load_checkpoint_years()

    for year in ALL_YEARS:
        if year in completed_years:
            continue

        try:
            raw_img = ee.Image(f"{GHS_BASE_ASSET_ID}{year}").select(BAND)
            img = raw_img.toDouble().where(raw_img.lt(0), 0).unmask(0)

            scale = img.projection().nominalScale().getInfo()

            combined = img.rename("PopT")
            for radius in BUFFER_DISTANCES:
                r_km = radius // 1000
                combined = combined.addBands(
                    img.updateMask(buffer_masks[radius]).rename(f"PopA_{r_km}")
                )

            result_fc = combined.reduceRegions(
                collection=fc_countries,
                reducer=ee.Reducer.sum(),
                scale=scale,
                tileScale=16,
            )
            results = result_fc.getInfo()

            year_records = []
            for feat in results["features"]:
                props = feat["properties"]
                country_name = props.get(COUNTRY_COLUMN)
                if country_name not in COUNTRY_LIST:
                    continue
                pop_t = max(0.0, float(props.get("PopT") or 0.0))
                for radius in BUFFER_DISTANCES:
                    r_km = radius // 1000
                    pop_a = max(0.0, float(props.get(f"PopA_{r_km}") or 0.0))
                    year_records.append({
                        "Country": country_name,
                        "Year": year,
                        "Radius": r_km,
                        "PopA": pop_a,
                        "PopT": pop_t,
                    })

            df_new = pd.DataFrame(year_records)
            file_exists = os.path.exists(CHECKPOINT_CSV)
            df_new.to_csv(CHECKPOINT_CSV, mode="a", header=not file_exists, index=False)

            build_excel_from_csv()

        except Exception as e:
            print(f"Failed for year {year}: {e}")
            time.sleep(3)


if __name__ == "__main__":
    run_analysis()
```

# Volcanoes

## Point

```python
PROJECT_ID = "your-gee-project-id"

VOLCANO_CSV_PATH = "data/volcanoes_africa.csv"
OUTPUT_FILE = "output/volcano.xlsx"
CHECKPOINT_CSV = "output/volcano_checkpoint.csv"

GHS_BASE_ASSET_ID = f"projects/{PROJECT_ID}/assets/WGS_"
BAND = "b1"
BUFFER_DISTANCES = [10000, 30000, 100000]
ALL_YEARS = list(range(1975, 2031, 5)) + list(range(2040, 2101, 10))


def load_volcanoes(csv_path):
    with open(csv_path, "r", encoding="utf-8-sig") as f:
        first_line = f.readline()
    delimiter = ";" if first_line.count(";") > first_line.count(",") else ","
    df = pd.read_csv(csv_path, delimiter=delimiter, encoding="utf-8-sig")
    df.columns = df.columns.str.strip()
    df = df.dropna(subset=["Latitude", "Longitude"])

    name_col = next((c for c in ["Volcano Name", "Volcano Na", "V_Name", "name"] if c in df.columns), None)

    features = []
    for _, row in df.iterrows():
        name = str(row[name_col]) if name_col else "Unknown"
        features.append(ee.Feature(
            ee.Geometry.Point([
                float(str(row["Longitude"]).replace(",", ".")),
                float(str(row["Latitude"]).replace(",", "."))
            ]),
            {"volcano_name": name}
        ))
    return features


def build_excel(checkpoint_path, output_path):
    df = pd.read_csv(checkpoint_path)
    if df.empty:
        return

    with pd.ExcelWriter(output_path, engine="xlsxwriter") as writer:
        for r_km in sorted(df["Radius"].unique()):
            sub = df[df["Radius"] == r_km]
            pivot = sub.pivot_table(
                index="Volcano", columns="Year", values="Value", aggfunc="mean"
            ).reset_index()
            pivot.columns.name = None
            sort_col = 2020 if 2020 in pivot.columns else pivot.columns[-1]
            pivot.sort_values(by=sort_col, ascending=False).to_excel(
                writer, sheet_name=f"Radius_{r_km}km", index=False
            )
        for sheet in writer.sheets.values():
            sheet.set_column("A:A", 30)
            sheet.set_column("B:Z", 12)


def load_checkpoint(volcano_count):
    completed = set()
    if os.path.exists(CHECKPOINT_CSV):
        df_ck = pd.read_csv(CHECKPOINT_CSV)
        for _, row in df_ck[["Year", "Radius"]].drop_duplicates().iterrows():
            yr_count = df_ck[
                (df_ck["Year"] == row["Year"]) & (df_ck["Radius"] == row["Radius"])
            ]["Volcano"].nunique()
            if yr_count >= volcano_count:
                completed.add((int(row["Year"]), int(row["Radius"])))
    return completed


def run_analysis():
    ee.Initialize(project=PROJECT_ID)

    raw_features = load_volcanoes(VOLCANO_CSV_PATH)
    volcano_names = [f.get("volcano_name").getInfo() for f in raw_features]

    buffered_fcs = {}
    for radius in BUFFER_DISTANCES:
        r_km = radius // 1000
        buffered = [
            ee.Feature(ee.Feature(f).geometry().buffer(radius), {"volcano_name": name})
            for f, name in zip(raw_features, volcano_names)
        ]
        buffered_fcs[r_km] = ee.FeatureCollection(buffered)

    completed = load_checkpoint(len(volcano_names))

    for year in ALL_YEARS:
        raw_img = ee.Image(f"{GHS_BASE_ASSET_ID}{year}").select(BAND)
        img = raw_img.toDouble().where(raw_img.lt(0), 0).unmask(0)
        scale = img.projection().nominalScale().getInfo()

        for radius in BUFFER_DISTANCES:
            r_km = radius // 1000
            if (year, r_km) in completed:
                continue

            try:
                result_fc = img.reduceRegions(
                    collection=buffered_fcs[r_km],
                    reducer=ee.Reducer.sum(),
                    scale=scale,
                    tileScale=16,
                )
                features = result_fc.getInfo()["features"]

                records = []
                for feat in features:
                    props = feat["properties"]
                    records.append({
                        "Volcano": props.get("volcano_name", "Unknown"),
                        "Year": year,
                        "Radius": r_km,
                        "Value": max(0.0, float(props.get("sum") or 0.0)),
                    })

                df_new = pd.DataFrame(records)
                file_exists = os.path.exists(CHECKPOINT_CSV)
                df_new.to_csv(CHECKPOINT_CSV, mode="a", header=not file_exists, index=False)
                completed.add((year, r_km))

            except Exception as e:
                print(f"Failed [{year}] {r_km}km: {e}")
                time.sleep(3)

    if os.path.exists(CHECKPOINT_CSV):
        build_excel(CHECKPOINT_CSV, OUTPUT_FILE)


if __name__ == "__main__":
    run_analysis()
```

## PEI Point

```python
FILE_PATH = "output/volcano.xlsx"
OUTPUT_PATH = "output/volcano_pei.xlsx"

WEIGHT_10KM = 0.941
WEIGHT_30KM = 0.057
WEIGHT_100KM = 0.002


def get_pei_score(weighted_pop):
    if weighted_pop <= 0:
        return 1
    elif weighted_pop < 3000:
        return 2
    elif weighted_pop < 10000:
        return 3
    elif weighted_pop < 30000:
        return 4
    elif weighted_pop < 100000:
        return 5
    elif weighted_pop < 300000:
        return 6
    else:
        return 7


def calculate_pei():
    df10 = pd.read_excel(FILE_PATH, sheet_name="Radius_10km").sort_values("Volcano").reset_index(drop=True)
    df30 = pd.read_excel(FILE_PATH, sheet_name="Radius_30km").sort_values("Volcano").reset_index(drop=True)
    df100 = pd.read_excel(FILE_PATH, sheet_name="Radius_100km").sort_values("Volcano").reset_index(drop=True)

    exclude = ["Volcano", "Country"]
    years = [col for col in df10.columns if col not in exclude]

    weighted_pop_df = df10[["Volcano"]].copy()
    pei_scores_df = df10[["Volcano"]].copy()

    for year in years:
        p10 = pd.to_numeric(df10[year], errors="coerce").fillna(0)
        p30 = pd.to_numeric(df30[year], errors="coerce").fillna(0)
        p100 = pd.to_numeric(df100[year], errors="coerce").fillna(0)

        w_sum = (p10 * WEIGHT_10KM) + (p30 * WEIGHT_30KM) + (p100 * WEIGHT_100KM)

        weighted_pop_df[year] = w_sum.round(0)
        pei_scores_df[year] = w_sum.apply(get_pei_score)

    with pd.ExcelWriter(OUTPUT_PATH, engine="xlsxwriter") as writer:
        pei_scores_df.to_excel(writer, sheet_name="PEI_Scores", index=False)
        weighted_pop_df.to_excel(writer, sheet_name="Weighted_Population", index=False)


if __name__ == "__main__":
    calculate_pei()
```

## Polygon

```python
PROJECT_ID = "your-gee-project-id"
VOLCANO_ASSET_PATH = f"projects/{PROJECT_ID}/assets/volcano_polygons"
OUTPUT_FILE = "output/volcano_polygons.xlsx"
CHECKPOINT_CSV = "output/volcano_polygons_checkpoint.csv"

GHS_BASE_ASSET_ID = f"projects/{PROJECT_ID}/assets/WGS_"
BAND = "b1"
BUFFER_DISTANCES = [10000, 30000, 100000]
ALL_YEARS = list(range(1975, 2031, 5)) + list(range(2040, 2101, 10))


def build_excel(checkpoint_path, output_path):
    df = pd.read_csv(checkpoint_path)
    if df.empty:
        return

    with pd.ExcelWriter(output_path, engine="xlsxwriter") as writer:
        for r_km in sorted(df["Radius"].unique()):
            sub = df[df["Radius"] == r_km]
            pivot = sub.pivot_table(
                index="Volcano", columns="Year", values="Value", aggfunc="mean"
            ).reset_index()
            pivot.columns.name = None
            sort_col = 2020 if 2020 in pivot.columns else pivot.columns[-1]
            pivot.sort_values(by=sort_col, ascending=False).to_excel(
                writer, sheet_name=f"Buffer_{r_km}km", index=False
            )
        for sheet in writer.sheets.values():
            sheet.set_column("A:A", 30)
            sheet.set_column("B:Z", 12)


def load_volcano_hulls(asset_path):
    volcano_fc = ee.FeatureCollection(asset_path)
    asset_features = volcano_fc.getInfo()["features"]

    volcano_names = []
    hull_features = []
    for feat in asset_features:
        props = feat["properties"]
        v_name = (props.get("Name") or props.get("name") or
                  props.get("V_NAME") or "Unknown")
        hull = ee.Feature(feat).geometry().convexHull(maxError=100)
        volcano_names.append(v_name)
        hull_features.append(ee.Feature(hull, {"volcano_name": v_name}))

    return volcano_names, hull_features


def load_checkpoint(volcano_count):
    completed = set()
    if os.path.exists(CHECKPOINT_CSV):
        df_ck = pd.read_csv(CHECKPOINT_CSV)
        for _, row in df_ck[["Year", "Radius"]].drop_duplicates().iterrows():
            n = df_ck[
                (df_ck["Year"] == row["Year"]) & (df_ck["Radius"] == row["Radius"])
            ]["Volcano"].nunique()
            if n >= volcano_count:
                completed.add((int(row["Year"]), int(row["Radius"])))
    return completed


def run_analysis():
    ee.Initialize(project=PROJECT_ID)

    volcano_names, hull_features = load_volcano_hulls(VOLCANO_ASSET_PATH)

    buffered_fcs = {}
    for radius in BUFFER_DISTANCES:
        r_km = radius // 1000
        buffered = [
            ee.Feature(feat.geometry().buffer(radius), {"volcano_name": name})
            for feat, name in zip(hull_features, volcano_names)
        ]
        buffered_fcs[r_km] = ee.FeatureCollection(buffered)

    completed = load_checkpoint(len(volcano_names))

    for year in ALL_YEARS:
        raw_img = ee.Image(f"{GHS_BASE_ASSET_ID}{year}").select(BAND)
        img = raw_img.toDouble().where(raw_img.lt(0), 0).unmask(0)
        scale = img.projection().nominalScale().getInfo()

        for radius in BUFFER_DISTANCES:
            r_km = radius // 1000
            if (year, r_km) in completed:
                continue

            try:
                result_fc = img.reduceRegions(
                    collection=buffered_fcs[r_km],
                    reducer=ee.Reducer.sum(),
                    scale=scale,
                    tileScale=16,
                )
                features = result_fc.getInfo()["features"]

                records = []
                for feat in features:
                    props = feat["properties"]
                    records.append({
                        "Volcano": props.get("volcano_name", "Unknown"),
                        "Year": year,
                        "Radius": r_km,
                        "Value": max(0.0, float(props.get("sum") or 0.0)),
                    })

                df_new = pd.DataFrame(records)
                file_exists = os.path.exists(CHECKPOINT_CSV)
                df_new.to_csv(CHECKPOINT_CSV, mode="a", header=not file_exists, index=False)
                completed.add((year, r_km))

            except Exception as e:
                print(f"Failed [{year}] {r_km}km: {e}")
                time.sleep(3)

    if os.path.exists(CHECKPOINT_CSV):
        build_excel(CHECKPOINT_CSV, OUTPUT_FILE)


if __name__ == "__main__":
    run_analysis()
```

# POPDENSITY

## Country point

```python
RING_STEP_KM = 2
MAX_RING_KM = 50
YEAR = 2000

PROJECT_ID = "your-gee-project-id"
VOLCANO_CSV_PATH = "data/volcanoes_africa.csv"
ASSET_STD = f"projects/{PROJECT_ID}/assets/GADM_410"
COUNTRY_COLUMN = "COUNTRY"
POP_ASSET_PREFIX = f"projects/{PROJECT_ID}/assets/WGS_"
POP_ASSET = f"{POP_ASSET_PREFIX}{YEAR}"
BAND = "b1"
OUTPUT_FILE = f"output/ring_density_by_country_point_{YEAR}.xlsx"

COUNTRY_LIST = sorted([
    "Cabo Verde", "Cameroon", "Chad", "Comoros", "Democratic Republic of the Congo",
    "Djibouti", "Equatorial Guinea", "Eritrea", "Ethiopia", "Mayotte",
    "Kenya", "Malawi", "Niger", "Nigeria", "Rwanda", "South Sudan", "Sudan",
    "Uganda", "Tanzania", "Zambia",
])


def load_coords_from_csv(csv_path):
    with open(csv_path, "r", encoding="utf-8-sig") as f:
        first_line = f.readline()
    delimiter = ";" if first_line.count(";") > first_line.count(",") else ","
    df = pd.read_csv(csv_path, delimiter=delimiter, encoding="utf-8-sig")
    df.columns = df.columns.str.strip()
    df = df.dropna(subset=["Latitude", "Longitude"])
    coords = []
    for _, row in df.iterrows():
        lon = float(str(row["Longitude"]).replace(",", "."))
        lat = float(str(row["Latitude"]).replace(",", "."))
        coords.append([lon, lat])
    return coords


def main():
    ee.Initialize(project=PROJECT_ID)

    coords = load_coords_from_csv(VOLCANO_CSV_PATH)
    multipoint = ee.Geometry.MultiPoint(coords)

    lons = [c[0] for c in coords]
    lats = [c[1] for c in coords]
    margin_deg = (MAX_RING_KM / 111.0) * 1.2
    study_bbox = ee.Geometry.Rectangle(
        [min(lons) - margin_deg, min(lats) - margin_deg,
         max(lons) + margin_deg, max(lats) + margin_deg],
        proj="EPSG:4326", geodesic=False
    )

    pop_img = ee.Image(POP_ASSET).select(BAND).toDouble()
    pop_img = pop_img.where(pop_img.lt(0), 0).unmask(0)
    scale = pop_img.projection().nominalScale().getInfo()

    countries_fc = ee.FeatureCollection(ASSET_STD)

    relevant_fc = (countries_fc
                   .filter(ee.Filter.inList(COUNTRY_COLUMN, COUNTRY_LIST))
                   .filterBounds(study_bbox))
    relevant_names = relevant_fc.aggregate_array(COUNTRY_COLUMN).getInfo()
    skipped = [c for c in COUNTRY_LIST if c not in relevant_names]

    country_geoms = {
        c: relevant_fc.filter(ee.Filter.eq(COUNTRY_COLUMN, c)).geometry(maxError=100)
        for c in relevant_names
    }
    all_countries_geom = relevant_fc.geometry(maxError=100)

    ring_defs = [
        (i, i + RING_STEP_KM, f"{i}-{i + RING_STEP_KM}km")
        for i in range(0, MAX_RING_KM, RING_STEP_KM)
    ]

    density_rows = {}
    area_rows = {}

    for inner_km, outer_km, label in ring_defs:
        ring_geom = (multipoint.buffer(outer_km * 1000) if inner_km == 0
                     else multipoint.buffer(outer_km * 1000).difference(
                         multipoint.buffer(inner_km * 1000), maxError=100))

        clip_features = []
        for c, geom in country_geoms.items():
            clip_geom = ring_geom.intersection(geom, maxError=100)
            clip_features.append(ee.Feature(clip_geom, {"country": c})
                                  .set("area_km2", clip_geom.area(maxError=100).divide(1e6)))
        total_clip = ring_geom.intersection(all_countries_geom, maxError=100)
        clip_features.append(ee.Feature(total_clip, {"country": "All countries"})
                              .set("area_km2", total_clip.area(maxError=100).divide(1e6)))

        flat_fc = ee.FeatureCollection(clip_features)

        result_fc = pop_img.reduceRegions(
            collection=flat_fc,
            reducer=ee.Reducer.sum(),
            scale=scale,
            tileScale=16,
        ).select(["country", "area_km2", "sum"])

        results = result_fc.getInfo()["features"]

        density_rows[label] = {"Ring": label}
        area_rows[label] = {"Ring": label}
        for feat in results:
            props = feat["properties"]
            country = props["country"]
            pop_sum = max(0.0, float(props.get("sum") or 0.0))
            area_km2 = max(0.0, float(props.get("area_km2") or 0.0))
            density = round(pop_sum / area_km2, 4) if area_km2 > 0 else 0.0
            density_rows[label][country] = density
            area_rows[label][country] = round(area_km2, 4)

        for c in skipped:
            density_rows[label][c] = 0.0
            area_rows[label][c] = 0.0

    ring_order = [label for _, _, label in ring_defs]
    col_order = ["Ring"] + COUNTRY_LIST + ["All countries"]

    with pd.ExcelWriter(OUTPUT_FILE, engine="xlsxwriter") as writer:
        for kind, rows, sheet in [("density", density_rows, "Density"), ("area", area_rows, "Area")]:
            df = pd.DataFrame([rows[r] for r in ring_order])
            df = df.reindex(columns=col_order)
            df.to_excel(writer, sheet_name=sheet, index=False)
            writer.sheets[sheet].set_column("A:A", 12)
            writer.sheets[sheet].set_column("B:Z", 14)


if __name__ == "__main__":
    main()
```

## Country polygon

```python
BUFFER_STEP_KM = 2
MAX_BUFFER_KM = 50
YEAR = 2000

PROJECT_ID = "your-gee-project-id"
VOLCANO_ASSET_PATH = f"projects/{PROJECT_ID}/assets/volcano_polygons"
ASSET_STD = f"projects/{PROJECT_ID}/assets/GADM_410"
COUNTRY_COLUMN = "COUNTRY"
GHS_BASE_ASSET_ID = f"projects/{PROJECT_ID}/assets/WGS_"
POP_ASSET = f"{GHS_BASE_ASSET_ID}{YEAR}"
BAND = "b1"
OUTPUT_FILE = f"output/ring_density_by_country_polygon_{YEAR}.xlsx"

MAX_RETRIES = 3
RETRY_WAIT = 10

COUNTRY_LIST = sorted([
    "Cabo Verde", "Cameroon", "Chad", "Comoros", "Democratic Republic of the Congo",
    "Djibouti", "Equatorial Guinea", "Eritrea", "Ethiopia", "Mayotte",
    "Kenya", "Malawi", "Niger", "Nigeria", "Rwanda", "South Sudan", "Sudan",
    "Uganda", "Tanzania", "Zambia",
])


def _extract_coords(geom_json):
    coords = []

    def walk(node):
        if isinstance(node, (int, float)):
            return
        if len(node) == 2 and all(isinstance(v, (int, float)) for v in node):
            coords.append(node)
            return
        for child in node:
            walk(child)

    walk(geom_json["coordinates"])
    return coords


def perform_analysis():
    ee.Initialize(project=PROJECT_ID)

    volcano_fc = ee.FeatureCollection(VOLCANO_ASSET_PATH)
    countries_fc = ee.FeatureCollection(ASSET_STD)

    asset_features = volcano_fc.getInfo()["features"]
    hulls = [ee.Feature(f).geometry().convexHull(maxError=100) for f in asset_features]

    all_coords = []
    for f in asset_features:
        all_coords.extend(_extract_coords(f["geometry"]))
    lons = [c[0] for c in all_coords]
    lats = [c[1] for c in all_coords]
    margin_deg = (MAX_BUFFER_KM / 111.0) * 1.2
    study_bbox = ee.Geometry.Rectangle(
        [min(lons) - margin_deg, min(lats) - margin_deg,
         max(lons) + margin_deg, max(lats) + margin_deg],
        proj="EPSG:4326", geodesic=False
    )

    gadm_names = set(countries_fc.aggregate_array(COUNTRY_COLUMN).getInfo())
    missing = [c for c in COUNTRY_LIST if c not in gadm_names]
    if missing:
        raise ValueError(f"Update COUNTRY_LIST, no GADM match for: {missing}")

    country_geoms = {
        c: countries_fc.filter(ee.Filter.eq(COUNTRY_COLUMN, c))
                        .geometry(maxError=100)
                        .intersection(study_bbox, maxError=100)
        for c in COUNTRY_LIST
    }
    all_countries_geom = countries_fc.filter(
        ee.Filter.inList(COUNTRY_COLUMN, COUNTRY_LIST)
    ).geometry(maxError=100).intersection(study_bbox, maxError=100)

    ring_defs = [(0, 0, "Hull")] + [
        (i, i + BUFFER_STEP_KM, f"{i}-{i + BUFFER_STEP_KM}km")
        for i in range(0, MAX_BUFFER_KM, BUFFER_STEP_KM)
    ]

    dissolved_rings = {}
    for inner_km, outer_km, label in ring_defs:
        if label == "Hull":
            ring_geoms = hulls
        elif inner_km == 0:
            ring_geoms = [h.buffer(outer_km * 1000) for h in hulls]
        else:
            ring_geoms = [
                h.buffer(outer_km * 1000).difference(h.buffer(inner_km * 1000), maxError=100)
                for h in hulls
            ]
        ring_fc = ee.FeatureCollection([ee.Feature(g) for g in ring_geoms])
        dissolved_rings[label] = ring_fc.geometry(maxError=100)

    pop_img = ee.Image(POP_ASSET).select(BAND).toDouble()
    pop_img = pop_img.where(pop_img.lt(0), 0).unmask(0)
    scale = pop_img.projection().nominalScale().getInfo()

    density_rows = {}
    area_rows = {}

    for label, dissolved_geom in dissolved_rings.items():
        clip_features = []
        for country in COUNTRY_LIST:
            clip_geom = dissolved_geom.intersection(country_geoms[country], maxError=100)
            clip_features.append(ee.Feature(clip_geom, {"country": country})
                                  .set("area_km2", clip_geom.area(maxError=100).divide(1e6)))
        total_geom = dissolved_geom.intersection(all_countries_geom, maxError=100)
        clip_features.append(ee.Feature(total_geom, {"country": "All countries"})
                              .set("area_km2", total_geom.area(maxError=100).divide(1e6)))

        flat_fc = ee.FeatureCollection(clip_features)

        for attempt in range(1, MAX_RETRIES + 1):
            try:
                result_fc = pop_img.reduceRegions(
                    collection=flat_fc,
                    reducer=ee.Reducer.sum(),
                    scale=scale,
                    tileScale=16,
                ).select(["country", "area_km2", "sum"])

                results = result_fc.getInfo()["features"]

                density_rows[label] = {"Ring": label}
                area_rows[label] = {"Ring": label}
                for feat in results:
                    props = feat["properties"]
                    country = props["country"]
                    pop_sum = max(0.0, float(props.get("sum") or 0.0))
                    area_km2 = max(0.0, float(props.get("area_km2") or 0.0))
                    density = round(pop_sum / area_km2, 4) if area_km2 > 0 else 0.0
                    density_rows[label][country] = density
                    area_rows[label][country] = round(area_km2, 4)

                break

            except Exception as e:
                if attempt < MAX_RETRIES:
                    time.sleep(RETRY_WAIT)
                else:
                    print(f"Ring {label} failed after {MAX_RETRIES} attempts: {e}")

    ring_order = [label for _, _, label in ring_defs]
    col_order = ["Ring"] + COUNTRY_LIST + ["All countries"]

    with pd.ExcelWriter(OUTPUT_FILE, engine="xlsxwriter") as writer:
        for kind, rows, sheet in [("density", density_rows, "Density"), ("area", area_rows, "Area")]:
            df = pd.DataFrame([rows[r] for r in ring_order if r in rows])
            df = df.reindex(columns=col_order)
            df.to_excel(writer, sheet_name=sheet, index=False)
            writer.sheets[sheet].set_column("A:A", 12)
            writer.sheets[sheet].set_column("B:Z", 14)


if __name__ == "__main__":
    perform_analysis()
```

## Volcano point

```python
YEAR = 2000
BUFFER_STEP_KM = 2
MAX_BUFFER_KM = 50

PROJECT_ID = "your-gee-project-id"
VOLCANO_CSV_PATH = "data/volcanoes_africa.csv"
POP_ASSET_PREFIX = f"projects/{PROJECT_ID}/assets/WGS_"
POP_ASSET = f"{POP_ASSET_PREFIX}{YEAR}"
OUTPUT_FILE = f"output/volcano_popdensity_point_{YEAR}_{BUFFER_STEP_KM}.xlsx"

BATCH_SIZE = 20
MAX_RETRIES = 3
RETRY_WAIT = 10


def load_volcanoes_from_csv(csv_path):
    with open(csv_path, "r", encoding="utf-8-sig") as f:
        first_line = f.readline()
    delimiter = ";" if first_line.count(";") > first_line.count(",") else ","
    df = pd.read_csv(csv_path, delimiter=delimiter, encoding="utf-8-sig")
    df.columns = df.columns.str.strip()
    df = df.dropna(subset=["Latitude", "Longitude"])
    volcanoes = []
    for _, row in df.iterrows():
        name = str(row.get("Volcano Name", row.get("V_Name", "Unknown")))
        lon = float(str(row["Longitude"]).replace(",", "."))
        lat = float(str(row["Latitude"]).replace(",", "."))
        volcanoes.append({"name": name, "lon": lon, "lat": lat})
    return volcanoes


def perform_analysis():
    ee.Initialize(project=PROJECT_ID)

    pop_img = ee.Image(POP_ASSET).select("b1").toDouble()
    pop_img = pop_img.where(pop_img.lt(0), 0).unmask(0)
    scale = pop_img.projection().nominalScale().getInfo()

    volcanoes = load_volcanoes_from_csv(VOLCANO_CSV_PATH)
    n = len(volcanoes)

    buffer_rings = [
        (i, i + BUFFER_STEP_KM, f"{i}-{i + BUFFER_STEP_KM}km")
        for i in range(0, MAX_BUFFER_KM, BUFFER_STEP_KM)
    ]

    density_rows = {}
    area_rows = {}

    for batch_start in range(0, n, BATCH_SIZE):
        batch = volcanoes[batch_start: batch_start + BATCH_SIZE]

        batch_features = []
        for v in batch:
            point = ee.Geometry.Point([v["lon"], v["lat"]])

            for inner_km, outer_km, label in buffer_rings:
                outer_geom = point.buffer(outer_km * 1000)
                ring_geom = (outer_geom if inner_km == 0
                             else outer_geom.difference(
                                 point.buffer(inner_km * 1000), maxError=100))
                batch_features.append(ee.Feature(ring_geom, {
                    "volcano_name": v["name"],
                    "ring_label": label,
                }).set("area_km2", ring_geom.area(maxError=100).divide(1e6)))

        flat_fc = ee.FeatureCollection(batch_features)

        for attempt in range(1, MAX_RETRIES + 1):
            try:
                result_fc = pop_img.reduceRegions(
                    collection=flat_fc,
                    reducer=ee.Reducer.sum(),
                    scale=scale,
                    tileScale=16,
                ).select(["volcano_name", "ring_label", "area_km2", "sum"])

                results = result_fc.getInfo()["features"]

                for feat in results:
                    props = feat["properties"]
                    v_name = props["volcano_name"]
                    label = props["ring_label"]
                    pop_sum = max(0.0, float(props.get("sum") or 0.0))
                    area_km2 = max(0.0, float(props.get("area_km2") or 0.0))
                    density = round(pop_sum / area_km2, 4) if area_km2 > 0 else 0.0

                    if v_name not in density_rows:
                        density_rows[v_name] = {"Volcano": v_name}
                        area_rows[v_name] = {"Volcano": v_name}

                    density_rows[v_name][label] = density
                    area_rows[v_name][label] = round(area_km2, 4)

                break

            except Exception as e:
                if attempt < MAX_RETRIES:
                    time.sleep(RETRY_WAIT)
                else:
                    print(f"Batch starting at {batch_start} failed after {MAX_RETRIES} attempts: {e}")

    col_order = ["Volcano"] + [label for _, _, label in buffer_rings]
    df_density = pd.DataFrame(list(density_rows.values()))[col_order]
    df_area = pd.DataFrame(list(area_rows.values()))[col_order]

    with pd.ExcelWriter(OUTPUT_FILE, engine="xlsxwriter") as writer:
        for df, sheet_name in [
            (df_density, "PopDensity_point"),
            (df_area, "SurfaceArea_km2"),
        ]:
            df.to_excel(writer, sheet_name=sheet_name, index=False)
            writer.sheets[sheet_name].set_column("A:A", 30)
            writer.sheets[sheet_name].set_column("B:Z", 15)


if __name__ == "__main__":
    perform_analysis()
```

## Volcano polygon

```python
YEAR = 2000
BUFFER_STEP_KM = 2
MAX_BUFFER_KM = 50

PROJECT_ID = "your-gee-project-id"
VOLCANO_ASSET_PATH = f"projects/{PROJECT_ID}/assets/volcano_polygons"
POP_ASSET_PREFIX = f"projects/{PROJECT_ID}/assets/WGS_"
POP_ASSET = f"{POP_ASSET_PREFIX}{YEAR}"
OUTPUT_FILE = f"output/volcano_popdensity_polygon_{YEAR}_{BUFFER_STEP_KM}.xlsx"

BATCH_SIZE = 20
MAX_RETRIES = 3
RETRY_WAIT = 10


def perform_analysis():
    ee.Initialize(project=PROJECT_ID)

    pop_img = ee.Image(POP_ASSET).select("b1").toDouble()
    pop_img = pop_img.where(pop_img.lt(0), 0).unmask(0)
    scale = pop_img.projection().nominalScale().getInfo()

    volcano_fc = ee.FeatureCollection(VOLCANO_ASSET_PATH)
    asset_features = volcano_fc.getInfo()["features"]
    n = len(asset_features)

    buffer_rings = [
        (i, i + BUFFER_STEP_KM, f"{i}-{i + BUFFER_STEP_KM}km")
        for i in range(0, MAX_BUFFER_KM, BUFFER_STEP_KM)
    ]

    density_rows = {}
    area_rows = {}

    for batch_start in range(0, n, BATCH_SIZE):
        batch = asset_features[batch_start: batch_start + BATCH_SIZE]

        batch_features = []
        for feat in batch:
            props = feat["properties"]
            v_name = (props.get("Name") or props.get("name") or
                      props.get("V_NAME") or "Unknown")
            hull = ee.Feature(feat).geometry().convexHull(maxError=100)

            batch_features.append(ee.Feature(hull, {
                "volcano_name": v_name,
                "ring_label": "Hull",
            }).set("area_km2", hull.area(maxError=100).divide(1e6)))

            for inner_km, outer_km, label in buffer_rings:
                outer_geom = hull.buffer(outer_km * 1000)
                ring_geom = (outer_geom if inner_km == 0
                             else outer_geom.difference(
                                 hull.buffer(inner_km * 1000), maxError=100))
                batch_features.append(ee.Feature(ring_geom, {
                    "volcano_name": v_name,
                    "ring_label": label,
                }).set("area_km2", ring_geom.area(maxError=100).divide(1e6)))

        flat_fc = ee.FeatureCollection(batch_features)

        for attempt in range(1, MAX_RETRIES + 1):
            try:
                result_fc = pop_img.reduceRegions(
                    collection=flat_fc,
                    reducer=ee.Reducer.sum(),
                    scale=scale,
                    tileScale=16,
                ).select(["volcano_name", "ring_label", "area_km2", "sum"])

                results = result_fc.getInfo()["features"]

                for feat in results:
                    props = feat["properties"]
                    v_name = props["volcano_name"]
                    label = props["ring_label"]
                    pop_sum = max(0.0, float(props.get("sum") or 0.0))
                    area_km2 = max(0.0, float(props.get("area_km2") or 0.0))
                    density = round(pop_sum / area_km2, 4) if area_km2 > 0 else 0.0

                    if v_name not in density_rows:
                        density_rows[v_name] = {"Volcano": v_name}
                        area_rows[v_name] = {"Volcano": v_name}

                    density_rows[v_name][label] = density
                    area_rows[v_name][label] = round(area_km2, 4)

                break

            except Exception as e:
                if attempt < MAX_RETRIES:
                    time.sleep(RETRY_WAIT)
                else:
                    print(f"Batch starting at {batch_start} failed after {MAX_RETRIES} attempts: {e}")

    col_order = ["Volcano", "Hull"] + [label for _, _, label in buffer_rings]
    df_density = pd.DataFrame(list(density_rows.values()))[col_order]
    df_area = pd.DataFrame(list(area_rows.values()))[col_order]

    with pd.ExcelWriter(OUTPUT_FILE, engine="xlsxwriter") as writer:
        for df, sheet_name in [
            (df_density, "PopDensity"),
            (df_area, "SurfaceArea_km2"),
        ]:
            df.to_excel(writer, sheet_name=sheet_name, index=False)
            writer.sheets[sheet_name].set_column("A:A", 30)
            writer.sheets[sheet_name].set_column("B:Z", 15)


if __name__ == "__main__":
    perform_analysis()
```
